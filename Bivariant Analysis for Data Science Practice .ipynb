```python
# Bivariant Analysis for Age-specific Percentage of Population Reporting LTHC(s) vs Subregion
```

```python
g = sns.relplot(
    x='Age-specific percentage of population reporting LTHC(s)', 
    y='Subregion_class', 
    hue='Age group',
    height=15,
    aspect=1.4,
    s=80,
    data=df
)

g.fig.suptitle(
    'Age-specific Percentage of Population Reporting LTHC(s) vs Subregion',
    fontsize=16,
    y=1.02
)

g.set_axis_labels(
    'Age-specific Percentage of Population Reporting LTHC(s)',
    'Subregion'
)

plt.savefig(
    'Age-specific Percentage of Population Reporting LTHC(s) vs Subregion.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
# Bivariant Analysis for Sex vs Percentage Reporting Long-Term Health Conditions
```

```python
sns.barplot(
    data=df,
    x='Sex',
    y='Age-specific percentage of population reporting LTHC(s)'
)

plt.title('Sex vs Percentage Reporting Long-Term Health Conditions')
plt.savefig(
    'Sex vs Percentage Reporting Long-Term Health Conditions.png',
    dpi=300,
    bbox_inches='tight'
)
plt.show()
```

```python
#Bivariant Analysis for Long-term health condition (LTHC) by Country of birth of person
```

```python
pd.crosstab(
    df['Country of birth of person'],
    df['Long-term health condition (LTHC)']
).plot(
    kind='bar',
    figsize=(30, 10),
    width=0.8
)

plt.xlabel(
    'Country of birth of person',
    fontsize=14
)

plt.ylabel(
    'Number of people',
    fontsize=14
)

plt.title(
    'Long-term health condition (LTHC) by Country of birth of person',
    fontsize=18,
    pad=20
)

# Rotate country names so they don't overlap
plt.xticks(
    rotation=60,
    ha='right',
    fontsize=9
)

# Put the legend outside the graph
plt.legend(
    title='Long-term health condition (LTHC)',
    bbox_to_anchor=(1.02, 1),
    loc='upper left',
    fontsize=9
)

plt.tight_layout()

plt.savefig(
    'LTHC_by_Country_of_Birth.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```
